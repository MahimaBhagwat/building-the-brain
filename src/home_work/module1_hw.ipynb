1. Variables, Data Types & Dynamic Typing

```python
# Problem 1.1: Type Casting & Truthiness Evaluation

def func(raw_price: float, raw_quantity: int, raw_discount: int):
    total_cost = float(raw_price) * int(raw_quantity)

    if(int(raw_discount) > 0):
        total_cost = total_cost - (total_cost * int(raw_discount) / 100)
    else: 
        total_cost = total_cost
        
    print(f"total_cost is None -> {total_cost is None} (false -> total_cost is a real number)")
    print(f"not total_cost -> {not total_cost} (false -> total_cost is not 0 therfore trutiness is true)")


func("199.99", "5", "0")
```

2. String Formatting & Manipulation

```python
# Problem 2.1: Dynamic Prompt Generator

def generate_llm_prompt(system_role: str, user_query: str, temperature: float):
    user_query = user_query.strip()
   # temperature = "{:.2f}".format(temperature)
    system_role = system_role.upper()
    user_query = user_query.capitalize()
    
    print(f"[SYSTEM: {system_role} | TEMP: {temperature:.2f}] {user_query}")
    
generate_llm_prompt(" system assistant ", " what is rag?", 0.7)
```

3. Control Flow: If-Else & Loops

```python
# Problem 3.1: API Retry Counter with Exponential Backoff

responses = [500, 502, 503, 200, 404]
for i in responses:
    if(i == None or i < 100 or i > 599):
        continue
    elif(i == 500 or i == 502 or i == 503):
        print(f"API call failed with status code {i}. Retrying in {2 ** responses.index(i)} seconds...")
    elif(i == 200):
        print(f"API call succeeded with status code {i}.")
        break
    else:
        print(f"fatal error: API call failed with status code {i}. Stopping retries.")
        break
    
```

```python
# Problem 5.1: The Safe System Log Collector

def add_log_entry(event: str, level: str = "INFO", logs: list[str] = None) -> list[str]:
    if logs is None:
        logs = []
    log_entry = f"[{level}] {event}"
    logs.append(log_entry)
    
    return logs

print(add_log_entry("System started"))
    
    
```
