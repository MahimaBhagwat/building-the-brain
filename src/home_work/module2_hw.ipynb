1.2 Classes & Object-Oriented Programming - AI Model Class

```python
class AIModel:
    def __init__(self, name, provider, model_type):
        self.name = name
        self.provider = provider
        self.model_type = model_type

    def describe(self):
        print(f"Model Name: {self.name}")
        print(f"Provider: {self.provider}")
        print(f"Model Type: {self.model_type}\n")

obj1 = AIModel("gpt-5", "OpenAI", "LLM")
obj1.describe()
obj2 = AIModel("bert", "HuggingFace", "Transformer")
obj2.describe()
obj3 = AIModel("dalle", "OpenAI", "Image Generation")
obj3.describe()

```

1.3 Dynamic Instance Attributes & Lifecycle - AI Agent Session

```python
import gc
class AgentSession:
    def __init__(self, user_id):
        self.user_id = user_id
        
    def __del__(self):
        print(f"Object deleted: user_id={getattr(self, 'user_id', 'unknown')}")
    
    def start(self):
        self.model = "gpt-5"
        self.message_count = 0
        self.is_active = True
        print(f"Starting session for user {self.user_id}")
    
    def end(self):
        self.is_active = False
        print(f"Ending session for user {self.user_id}")
        

session = AgentSession("user_123")
print("Before start")

session.start()
print("After start")

session.end()
print("After end")

del session
gc.collect()

print("After del session")
```

1.4 Class Methods - AI Model Factory

```python
class AIModel:
    def __init__(self, name, provider):
        self.name = name
        self.provider = provider
    
    @classmethod
    def from_string(cls, s : str):
        name, provider = s.split(":")
        return cls(name, provider)

model = AIModel.from_string("GPT-5:OpenAI")
print(model.name)
print(model.provider)
```

1.5 Inheritance & super() - AI Engineer Hierarchy

```python
class SoftwareEngineer:
    def __init__(self, name, experience):
        self.name = name
        self.experience = experience
    
    def describe(self):
        print(f"Name: {self.name}, Experience: {self.experience} years")
        
class AIEngineer(SoftwareEngineer):
    def __init__(self, name, experience, specialization):
        super().__init__(name, experience)
        self.specialization = specialization

    def describe(self):
        super().describe()
        print(f"Specialization: {self.specialization}")

eng = AIEngineer("Alice", 5, "Machine Learning")
eng.describe()
```

2.2 Concurrency with asyncio.gather()

```python
import time
import asyncio

async def download_document():
    print(f"Downloading document from ")
    await asyncio.sleep(3)
    print("Download complete")

async def generate_embeddings():
    print("Generating embeddings")
    await asyncio.sleep(2)
    print("Embeddings generated")

async def fetch_metadata():
    print("Fetching metadata")
    await asyncio.sleep(1)  
    print("Metadata fetched")
    
async def main():
    start_time = time.perf_counter()

    await asyncio.gather(
        download_document(),
        generate_embeddings(),
        fetch_metadata(),
    )

    elapsed_time = time.perf_counter() - start_time
    print(f"Total execution time: {elapsed_time:.2f} seconds")


await main()
```

2.3  Concurrent API Requests

```python
import asyncio
import httpx

async def fetch_data(url):
    async with httpx.AsyncClient() as client:
        response = await client.get(url)
        return response.json()

async def main():
    user_url = "https://jsonplaceholder.typicode.com/users/1"
    post_url = "https://jsonplaceholder.typicode.com/posts/1"
    todo_url = "https://jsonplaceholder.typicode.com/todos/1"

    user, post, todo = await asyncio.gather(
        fetch_data(user_url),
        fetch_data(post_url),
        fetch_data(todo_url),
    )
    
    print(f"User: {user['name']}")
    print(f"Post ID: {post['id']}")
    print(f"Todo: {todo['title']}")

await main()
```

2.4 Build an Async API Client

```python
class PostService:
    def __init__(self, base_url):
        self.base_url = base_url
        
    async def get_post(self, post_id):
        url = f"{self.base_url}/posts/{post_id}"
        async with httpx.AsyncClient() as client:
                response = await client.get(url)
                response.raise_for_status()
                
                return response.json()

async def main():
    post_service = PostService("https://jsonplaceholder.typicode.com")
    try:
        id1, id2, id3 = await asyncio.gather(
            post_service.get_post(1),
            post_service.get_post(2),
            post_service.get_post(3),
        )
    except Exception as e:
        print(f"Error occurred: {e}")
    else :
        print(f"Post 1: {id1['title']}")
        print(f"Post 2: {id2['title']}")
        print(f"Post 3: {id3['title']}")
    
await main()
    
```

2.5 Build a Concurrent LLM-Style Data Fetcher

```python

import asyncio
import httpx
import time

class Fetcher:
    def __init__(self, base_url, client):
        self.base_url = base_url
        self.client = client

    async def get_user(self, user_id):
        url = f"{self.base_url}/users/{user_id}"
        response = await self.client.get(url)
        response.raise_for_status()
        return response.json()

    async def get_posts(self):
        url = f"{self.base_url}/posts"
        response = await self.client.get(url)
        response.raise_for_status()
        return response.json()

    async def get_todos(self):
        url = f"{self.base_url}/todos"
        response = await self.client.get(url)
        response.raise_for_status()
        return response.json()


async def main():
    start = time.perf_counter()

    async with httpx.AsyncClient() as client:
        fetcher = Fetcher("https://jsonplaceholder.typicode.com", client)

        user, posts, todos = await asyncio.gather(
            fetcher.get_user(1),
            fetcher.get_posts(),
            fetcher.get_todos()
        )

        result = {
            "user": user,
            "posts": posts,
            "todos": todos
        }

        print(result)

    end = time.perf_counter()
    print(f"\nConcurrent execution time: {end - start:.4f} seconds")

await main()
```

```python
from pydantic import BaseModel, EmailStr, SecretStr

class User(BaseModel):
    name: str
    age: int
    email: EmailStr
    phone: str = None 
    password : SecretStr
    
user = User(
    name="Mahima",
    age=21,
    email="mahima@example.com",
    phone="9876543210",
    password="mysecretpassword"
)

# Direct attribute access
print(user.name)
print(user.age)
print(user.email)
print(user.phone)

print(user.password)

print(user.password.get_secret_value())

# Convert into dictionary
user_dict = user.model_dump()

# Convert into JSON string
user_json = user.model_dump_json()

print(user_dict)
print(user_json)

# Deserialize JSON string into Pydantic object
new_user = User.model_validate_json(user_json)

print(new_user)
print(new_user.name)
print(type(new_user))
```

3.2 Nested Order Model

```python

from pydantic import BaseModel
from typing import List

class Customer(BaseModel):
    name: str
    email: str
    phone: str

class Address(BaseModel):
    city: str
    state: str
    pincode: int

class Product(BaseModel):
    product_name: str
    price: float
    quantity: int

class Order(BaseModel):
    order_id: int
    customer: Customer
    address: Address
    products: List[Product]


# Input data
order_data = {
    "order_id": 101,
    "customer": {
        "name": "Mahima",
        "email": "mahima@example.com",
        "phone": "9876543210"
    },
    "address": {
        "city": "Pune",
        "state": "Maharashtra",
        "pincode": 411001
    },
    "products": [
        {
            "product_name": "Laptop",
            "price": 65000,
            "quantity": 1
        },
        {
            "product_name": "Mouse",
            "price": 800,
            "quantity": 2
        }
    ]
}

# Validate input and create Pydantic object
order = Order.model_validate(order_data)

# Access nested fields using attributes
print(order.order_id)

print(order.customer.name)
print(order.customer.email)

print(order.address.city)
print(order.address.pincode)

# Access collection of products
for product in order.products:
    print(product.product_name)
    print(product.price)
    print(product.quantity)
```

3.3 Validated API Request

```python
from pydantic import Field, field_validator
from typing import Optional, Annotated

class UserRequest(BaseModel):
    # Required field with length constraints
    name: Annotated[str, Field(min_length=3, max_length=30)]

    # Required field with email validation
    email: EmailStr

    # Numeric constraints
    age: int = Field(ge=18, le=60)

    # Optional field
    phone: Optional[str] = None

    # Aliased field
    username: str = Field(alias="user_name", min_length=4, max_length=20)

    # Optional field with a default value
    city: str = "Pune"

    # Custom validator: business rule
    @field_validator("username")
    @classmethod
    def validate_username(cls, value):
        if not value.isalnum():
            raise ValueError("Username must contain only letters and numbers")
        return value


# Input data
data = {
    "name": "Mahima",
    "email": "mahima@example.com",
    "age": 21,
    "phone": "9876543210",
    "user_name": "mahima123"
}

# Validate input
user = UserRequest.model_validate(data)

# Access attributes
print(user.name)
print(user.email)
print(user.age)
print(user.phone)
print(user.username)
print(user.city)

# Serialize using aliases
print(user.model_dump(by_alias=True))
```

3.4 Application Configuration

```python
class AppConfig(BaseModel):
    app_name: str = Field(min_length=3, max_length=50)
    environment: str = Field(min_length=3, max_length=20)
    api_key: SecretStr

    model_config_data: dict = Field(
        default_factory = dict
    )
    
    description: Optional[str] = None
    max_tokens: int = Field(default=1000, ge=100, le=10000)


config = AppConfig(
    app_name="KnowledgeOS",
    environment="development",
    api_key="secret123",
    model_config_data={
        "model": "gemini-2.5-flash",
        "temperature": 0.7,
        "top_p": 0.9
    },
    description="Document retrieval application",
    max_tokens=2000
)

print(config.app_name)
print(config.environment)
print(config.api_key)
print(config.model_config_data)
print(config.max_tokens)

print(config.api_key.get_secret_value())

print(config.model_dump())

config_json = config.model_dump_json()
print(config_json)

new_config = AppConfig.model_validate_json(config_json)

print(new_config)
print(new_config.api_key)
```

3.5  LLM Response Pipeline

```python
from typing import Any, Dict

class LLMResponse(BaseModel):
    answer: str = Field(min_length=5, max_length=1000)

    confidence: float = Field(ge=0.0, le=1.0)

    sources: List[str] = Field(min_length=1)

    metadata: Dict[str, Any] = Field(default_factory=dict)
    
    @field_validator("sources")
    @classmethod
    def validate_sources(cls, value):
        for source in value:
            if not source.strip():
                raise ValueError("Source cannot be empty")
        return value
    
# 1. JSON Response
llm_json = """
{
    "answer": "Python supports asynchronous programming using async and await.",
    "confidence": 0.95,
    "sources": ["Python Documentation", "Asyncio Documentation"],
    "metadata": {
        "model": "gemini-2.5-flash",
        "tokens": 120,
        "response_time": 1.5
    }
}
"""

# 2. Pydantic Validation
response = LLMResponse.model_validate_json(llm_json)

# 3. Python Object
print(f"Validated Object: {response}")

# 4. Application Logic
if response.confidence >= 0.8:
    final_answer = response.answer
else:
    final_answer = "I'm not confident enough to answer."

# 5. Create final response
final_response = LLMResponse(
    answer=final_answer,
    confidence=response.confidence,
    sources=response.sources,
    metadata=response.metadata
)

# 6. Serialization
output_json = final_response.model_dump_json(indent=4)

print(f"\nFinal JSON:")
print(output_json)
```
