#### **Class and object**

- Avoid mutable class attributes when you want per-object state.
- @classmethod are mostly used as factory methods/ alternative constructors; it receives the class as its first argument, conventionally named cls. Use it when the method needs to read or change class-level data, or create instances.
- @staticmethod receives no automatic first argument—neither the instance nor the class. Use it for a helper function that belongs conceptually with the class but doesn’t need either.
- @property decorator makes a method readable like a plain attribute without parentheses. Helps in creating a dynamic attibutes from other attributes.
- Python has no private / protected access modifiers - only conventions, enforced by nobody(underscore : _name)
- A single leading underscore (_name ) signals "internal, don't touch this from outside" and
Python trusts you to respect it. A double leading underscore (__name ) triggers name mangling
(Python renames it internally to avoid subclass clashes) - know it exists, but a single underscore
is the idiomatic default, not name

````python
class Employee:
    # attribute belonging to the class
    company = "ABC Corp"
    
    def __init__(self, f_name: str, l_name: str):
        # attributes belonging to the instance
        self.first_name = f_name
        self.last_name = l_name
    
    def info(self) -> str:
        self.department = "Engineering"  # Adding a new attribute dynamically
        return f"Name: {self.first_name} {self.last_name}, Department: {self.department}"
    
    @classmethod
    def change_company(cls, name):
        cls.company = name
    
    
    @staticmethod
    def is_valid_name(name):
        return bool(name.strip())
    
    @property
    def full_name(self) -> str:
        return f"{self.first_name} {self.last_name}"
    # @property
    # def full_name(cls) -> str:
    #         return f"{cls.first_name} {cls.last_name}"
    
    
print(Employee.change_company("XYZ Inc"))  # Accessing class method without an instance; cls = class
print(Employee.is_valid_name("John"))  # Accessing static method without an instance; no cls or self is passed to the method

emp = Employee("John", "Doe")
# print(emp.department)  # raises AttributeError since 'department' is not defined yet
print(f"Employee info : {emp.info()}")
print(f"Employee department : {emp.department}")  # Now it works, 'department' is defined 

print(f"Employee full name : {emp.full_name}")  # not emp.full_name() because full_name is a property, not a method

# in python we can define attributes inside the constructor or anywhere we want inside the lifecycle of that particular object.
````

#### **Inheritance**

- No extends or ' : ' is needed for inheriting a class into other use **'child_class(base class)'**
- super() used to call base class methods

````python
class Manager(Employee):
    def __init__(self, f_name: str, l_name: str, team_size: int):
        super().__init__(f_name, l_name)
        self.team_size = team_size
    
    def info(self) -> str:
        # department is not defined in the constructor, so we cannot access it here. It is defined in the info method of Employee class, so we can access it after calling the info method of Employee class.
        # return f"Name: {self.first_name} {self.last_name}, Department: {self.department}, Team Size: {self.team_size}"
        return f"Name: {self.first_name} {self.last_name}, Team Size: {self.team_size}" 
    

emp1 = Employee("Alice", "Smith")
mgr1 = Manager("Bob", "Johnson", 5)
print(f"Employee info : {emp1.info()}")
print(f"Manager info : {mgr1.info()}")


````

#### **Composition and Duck Typing**

- Composition -> means to pass one class obj as a parameter to constructor of another class
- Duck typing -> means Python cares about what an object can do, rather than its exact class or declared type; it is useful for writing flexible code that works with different classes—without requiring them to share a parent class. If an object lacks the method your code tries to use, Python raises an error at runtime.

````python
class Tool:
    def __init__(self, name: str, price: float):
        self.name = name
        self.price = price
    
    def info(self) -> str:
        return f"Tool Name: {self.name}, Price: {self.price}"

class Agents:
    def __init__(self, name: str, tool: Tool):
        self.name = name
        self.tool = tool
    
    def info(self) -> str:
        return f"Agent Name: {self.name}, Tool: {self.tool.info()}"  # composition

tool = Tool("Web search", 100.0)
agent = Agents("GPT", tool)
print(f"\n{agent.info()}")  

# ---------------------------------------------------------------------------------------------------------------------

class Dog:
    def speak(self) -> str:
        return "Woof!"

class Cat:
    def speak(self) -> str:
        return "Meow!"
    
def make_sound(animal):
    print(animal.speak())
    
# here the fuction doesnt require a specific parent class, it simply aspects the object to provide speak().
# .speak() can be called on a Car() also if it has a speak() method.
````

#### **Dunder Methods**

They are special methods with double underscore.
Python calls them in response to common operations, letting your objects work naturally with built-in syntax and functions.
some examples :
- `__init__` — initializes an instance when you create it.
- `__str__` — provides a readable string for users, such as with `str(obj)` or `print(obj)`.
- `__repr__` — provides a useful representation for debugging.
- `__len__` — lets you use `len(obj)`.
- `__eq__` — defines how `obj1 == obj2` behaves.
- `__call__` — lets you call an object like a function: `obj()`.
- `__add__` — defines what happens when you use `+` with an object: `obj1 + obj2`.
- `__contains__` — defines membership checks using `in`: `"item" in obj`
- `__lt__` — defines “less than” comparisons using `<`, such as `obj1 < obj2`.




#### **Data Class**

#### **Context Manager**

handles setup and cleanup around a block of code.<br>
For example, opening a file with `with` ensures it is closed automatically—even if an error occurs:
```python
with open("example.txt", "r") as file:
    contents = file.read()
```

A context manager class typically defines two dunder methods:
- `__enter__` — runs when execution enters the `with` block.
- `__exit__` — runs when execution leaves the block, for cleanup.<br>

Use context managers when a resource needs to be reliably cleaned up, such as a file or a database connection.


#### **Iterables and Generators**

- iterable is an object you can loop over, such as a list, string, or tuple. It provides its items one at a time when you iterate over it.
- A generator is a convenient way to create an iterator that produces values on demand, using yield. This can save memory when working with large sequences.
    - Calling count_up_to(3) creates a generator; each iteration resumes the function and produces the next value.



````python
def count_up_to(limit):
    number = 1
    while number <= limit:
        yield number
        number += 1

for number in count_up_to(3):
    print(number)
````

#### **Variadic Arguments** 

Variadic arguments let a function accept a variable number of arguments.

- `*args` collects extra **positional arguments** into a tuple.
- `**kwargs` collects extra **keyword arguments** into a dictionary.

````python
def show_args(*args, **kwargs):
    print(args)
    print(kwargs)
    print(f"Name is: {kwargs.get('name', 'No name provided')}")

show_args("hello", 42, name="Alice", active=True)

# -------------------------------------------------------------------------------------------------------------------------------------
# Unpacking dict or tuple into function arguments using **(when there is mapping) and *(when there is a sequence/list) operators.

def greet(name, greeting):
    print(f"\n{greeting}, {name}!")

details = {"name": "Alice", "greeting": "Hello"}
# greet(details)
greet(**details)
````
