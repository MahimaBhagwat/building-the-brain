String formatting and different methods

```python
name = "Mahima"
greeting = f"Hello, {name}"  # f-string
print(greeting)  # Output: Hello, Mahima

paragraph = """This is a multi-line string
Basically a paragraph which can have multiple lines.

It can be used for documentation or long text."""

print(paragraph)

#newline character
print("\n")  

#String methods
print(f"strip() -> {'   Hello, World!   '.strip()}  -> strips leading and trailing whitespace")  # Output: Hello, World!
print(f"upper() -> {'hello, world!'.upper()}  -> converts to uppercase")  # Output: HELLO, WORLD!
print(f"lower() -> {'HELLO, WORLD!'.lower()}  -> converts to lowercase")  # Output: hello, world!
print(f"replace() -> {'Hello, World!'.replace('World', 'Python')}  -> replaces substring")  # Output: Hello, Python!
print(f"split() -> {'Hello, World!'.split(',')}  -> splits string into list of strings")  # Output: ['Hello', ' World!']
print(f"slice() -> {'Hello, World!'[7:12]}  -> slices string from index 7 to 11")  # Output: World

print(f"join() -> {', '.join(['Hello', 'World'])}  -> joins list into string")  # Output: Hello, World
print(f"startswith() -> {'Hello, World!'.startswith('Hello')}  -> checks if string starts with substring")  # Output: True
print(f"endswith() -> {'Hello, World!'.endswith('!')}  -> checks if string ends with substring")  # Output: True
print(f"find() -> {'Hello, World!'.find('World')}  -> finds substring index")  # Output: 7
print(f"count() -> {'Hello, World!'.count('o')}  -> counts occurrences of substring")  # Output: 2
print(f"format() -> {'Hello, {}!'.format('World')}  -> formats string with placeholders")  # Output: Hello, World!
print(f"len() -> {len('Hello, World!')}  -> returns length of string")  # Output: 13
print(f"capitalize() -> {'hello, world!'.capitalize()}  -> capitalizes first letter")  # Output: Hello, world!
print(f"title() -> {'hello, world!'.title()}  -> capitalizes first letter of each word")  # Output: Hello, World!
print(f"isdigit() -> {'12345'.isdigit()}  -> checks if string is numeric")  # Output: True
print(f"isalpha() -> {'Hello'.isalpha()}  -> checks if string is alphabetic")  # Output: True
print(f"isalnum() -> {'Hello123'.isalnum()}  -> checks if string is alphanumeric")  # Output: True
print(f"isspace() -> {'   '.isspace()}  -> checks if string is whitespace")  # Output: True

```

Loops

```python
users = ["mahima", "john", "alice", 1, False, 3.14]
users.append("bob")  # Add a new user to the list
print(users) 

print("for-each over a list")
for user in users:
    print(user)
    
print("\n")

print("iterative counting over a range")
for i in range(len(users)):
    print(f"User {i}: {users[i]}")
    
print("\n")

print("iterative counting over a range with step")
#for i in range(len(users), 2): --> This line seems to have a syntax error. The correct usage of range with a step would be range(start, stop, step). If you want to iterate over the users list with a step of 2, you can do it like this:
for i in range(0, len(users), 2):
    print(f"User {i}: {users[i]}")
    
print("\n")

print("while loop over a list")
i = 0
while i < len(users):
    print(f"User {i}: {users[i]}")
    i += 1

print("\n")

print("for loop with break")
for i in users:
    if i == "john":
        continue
    if i == "alice":
        print("Found Alice!")
        break
    print(f"processing user: {i}")
```

Functions and Type Hints

Important : Python type hints are not enforced at runtime by interpreter. Libraries like pydantic or mypy are used to perform runtime checking or static verification.

```python
def greet(name: str, greeting: str = "Hello") -> str:
    return f"{greeting}, {name}!"

print(greet("Mahima"))  # default greeting
print(greet("Mahima", "Hi"))  # Overriding default 
print(greet(name="Mahima", greeting="Hi"))  # keyword arguments

#type hints are optional in python, they are not enforced at runtime. They are mainly used for static type checking and code readability.
print(f"type hints are not enforced : {greet(123)}  (this works even though name is expected to be a string)")
```

```python
a = [1, 2, 3]
b = [2, 3, 4]
b = a
b.append(5)
print(f"a : {a}, b : {b}  (both a and b are same because b is assigned to a, so any change in b will reflect in a)")

# this is because lists are mutable objects in python, so when you assign b = a, both a and b point to the same list object in memory. Any changes made to that list through either variable will be reflected in the other.
# this does not happen with immutable objects like integers, strings, and tuples. When you assign a new value to an immutable object, a new object is created in memory, and the variable now points to that new object.
x = 5
y = x
x = 4
print(f"x : {x}, y : {y}  (x and y are different because y is assigned to x, so any change in x will not reflect in y)")
```

**Exception Handling**

```python
class APIException(Exception):
    """Custom exception for API errors."""
    pass

def process_api_payload(payload: dict):
    if "prompt" not in payload:
        raise APIException("Missing 'prompt' in payload.")
    else:
        print(f"Processing payload with prompt: {payload['prompt']}")
        
def execute_pipeline(payload: dict):
    try:
        result = process_api_payload(payload)
    except APIException as e :
        print(f"APIException occurred: {e}")
    except Exception as e :
        print(f"An unexpected error occurred: {e}")
    else :
        print(f"Pipeline executed successfully: {result}.")
    finally :
        print("Pipeline execution completed. Closing resources if any.")

print("------- Testing with invalid payload ---")
execute_pipeline({})

print("------- Testing with valid payload ---")
execute_pipeline({"prompt": "Hello, World!"})
```
